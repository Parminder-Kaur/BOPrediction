```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

df = pd.read_csv('tmdb_5000_movies.csv')

#Filter on Sci-Fi movies
Only_scifi = df[df['genres'].str.contains('Science Fiction', na=False)].copy()
X = df[['budget','popularity']] # Independent variables
y = df['revenue'] # dependent variables

X=X.dropna() # drop all NaN
y=y.loc[X.index] # alighn with x axis for right indexes

# train and test 
#test_size tells Python to take 20% of movies and hide them in a "vault" (X_test and y_test). The model never sees these during training.
#X_train, y_train are the other 80% of movies the model uses to learn the relationship between money and fame.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)


# Initialize and train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

#make predictions
y_pred = model.predict(X_test)

plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred, color='purple')
# y_test (X-axis): These are the actual box office numbers from the real world.
# y_pred (Y-axis): These are the predicted numbers your AI guessed.


plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2) # Line of perfect prediction

# make it look nice
plt.title('Sci-Fi Box Office: Actual vs. Predicted Revenue', fontsize=14)
plt.xlabel('Actual Revenue ($)', fontsize=12)
plt.ylabel('Predicted Revenue ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
